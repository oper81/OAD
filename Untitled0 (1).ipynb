```python
filtered_tracks = df[df['release_year'] > 2024]
display(filtered_tracks)
```

С помощью логического условия df['release_year'] > 2024 мы отфильтровали датасет, выбрав только те музыкальные композиции, год выпуска которых относится к 2025 году.

```python
popularity_stats = df['popularity'].describe()
print(popularity_stats)
```

```python
print(f"\nМедиана для popularity: {df['popularity'].median()}")
```

Функция describe() вывела базовые метрики для столбца popularity:
count — количество заполненных строк.
mean — среднее арифметическое значение популярности.
std — стандартное отклонение (мера разброса данных вокруг среднего). min — минимальное значение популярности в датасете.50% (или median) — медиана (значение, делящее выборку пополам).
max — максимальное значение популярности трека.

```python
mode_genre = df['genre'].mode()[0]
mode_year = df['release_year'].mode()[0]

print(f"Наиболее часто встречающийся жанр (мода): {mode_genre}")
print(f"Наиболее часто встречающийся год выпуска (мода): {int(mode_year)}")
```

С помощью метода mode() мы определили самые часто встречающиеся значения в таблице. Индекс [0] берется на случай, если самых популярных значений окажется несколько.

```python
plt.figure(figsize=(10, 6))
plt.hist(df['popularity'], bins=20, color='skyblue', edgecolor='black')
plt.title('Распределение значений популярности треков (Popularity)')
plt.xlabel('Уровень популярности')
plt.ylabel('Количество треков')
plt.grid(axis='y', alpha=0.5)
plt.show()
```

Построена гистограмма распределения уровня популярности. Ответы по графику: Основная часть значений популярности сосредоточена в диапазоне от 1000 до 1400. Значения на краях распределения (очень низкая популярность около 0 или максимальная около 1600) встречаются значительно реже остальных.

Что такое DataFrame?

DataFrame — это основная структура данных в библиотеке Pandas, которая представляет собой двумерную таблицу. У неё есть строки и именованные столбцы. Это прямой аналог обычной таблицы в Excel или SQL.

Что такое признак (Feature)?

Признак (или Feature) — это столбец (колонка) в таблице данных. Он отражает конкретное свойство, характеристику или измерение изучаемого объекта (в данном случае трека: например, artist_name, genre, popularity или tempo).

Чем среднее арифметическое отличается от медианы?

Среднее арифметическое — это сумма всех значений, деленная на их количество. Оно сильно зависит от аномально больших или маленьких значений (выбросов).

Медиана — это «середина» отсортированного списка чисел. Ровно половина данных в таблице больше этого значения, а половина — меньше. Медиана гораздо лучше показывает реальную картину, если в данных есть сильные перекосы и выбросы.

Что показывает гистограмма?

Гистограмма наглядно показывает распределение данных. Она делит весь диапазон значений (например, популярность от 0 до 100) на равные отрезки-интервалы и строит столбцы, высота которых отражает, как часто (сколько именно треков) попадает в каждый из этих интервалов.

Для чего нужна фильтрация данных?

Фильтрация нужна для того, чтобы отсечь лишнюю информацию и оставить в таблице только те строки, которые подходят под заданное условие (например, выбрать треки строго определенного года или жанра). Это позволяет сфокусироваться на изучении конкретной выборки.
